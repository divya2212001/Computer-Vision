```python
from ultralytics import YOLO
import glob, os

detect_model = YOLO("yolov8s.pt")

image_paths = sorted(glob.glob("frames/*.png"))

os.makedirs("detect_frames", exist_ok=True)

for img in image_paths:
    results = detect_model(img, verbose=False)
    results[0].save(
        filename=os.path.join(
            "detect_frames",
            os.path.basename(img)
        )
    )

print("Detection Completed")
```

```python
!ffmpeg -y -framerate 30 \
-i detect_frames/frame_%04d.png \
-c:v libx264 -pix_fmt yuv420p \
detected_video.mp4
```

```python
seg_model = YOLO("yolov8n-seg.pt")

os.makedirs("seg_frames", exist_ok=True)

for img in image_paths:
    results = seg_model(img, verbose=False)
    results[0].save(
        filename=os.path.join(
            "seg_frames",
            os.path.basename(img)
        )
    )

print("Segmentation Completed")
```

```python
!ffmpeg -y -framerate 30 \
-i seg_frames/frame_%04d.png \
-c:v libx264 -pix_fmt yuv420p \
segmented_video.mp4
```

```python
!ffmpeg -y -i raw.mp4 -an raw_noaudio.mp4
!ffmpeg -y -i detected_video.mp4 -an detect_noaudio.mp4
!ffmpeg -y -i segmented_video.mp4 -an seg_noaudio.mp4
```
